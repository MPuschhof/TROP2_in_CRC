## Visualize how datasets will be reconstructed

Use infer_study_config_from_files to see how the processed data files will be used to reconstruct an anndata object. An SingleCellExperiment object for visualization in R can be reassembled in analogous fashion from the processed files.

```python
# Check how LM3 has been assembled in E-MTAB-16836
accession = "E-MTAB-16836"
accession_dir = Path(download_dir) / accession

infer_study_config_from_files(
    accession_dir=accession_dir,
    dataset="LM3",
    on_multiple="error",
    verbose=True,
    blacklist=('')
)
```

## All-in-1 wrapper function

This all-in-1 wrapper takes care of downloading the data and reassembling an anndata object.

When multiple datasets (data subsets) are available in the same accession, the wrapper reconstructs all and loads one into memory. Here, a full dataset and a dataset subsetted to epithelial cells only are available. The latter is loaded per default.

```python
# All-in-1 wrapper to download data and reassemble anndata object
adata = get_public_study_adata(
    accession="E-MTAB-16583",
    download_dir=download_dir,
)
adata
```
